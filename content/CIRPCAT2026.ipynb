## An Algebra of Geometry

So far we have only discussed geometry with oriented planes, but done no algebra.
<div class="fragment">
But the rules I have explained are exactly the rules of Geometric Algebra (GA):
a Geometric (Clifford) Algebra $\mathbb{R}_{pqr}$ of $d$ = $p+q+r$ dimensions has $d$ anti-commuting ($\mathbf{e}_i \mathbf{e}_j=-\mathbf{e}_j \mathbf{e}_i$) basis vectors $\mathbf{e}_i$, of which $p$ are positive basis vectors $\mathbf{e}_i^2 = 1$, $q$ are negative basis vectors $\mathbf{e}_i^2 = -1$, and $r$ are null basis vectors $\mathbf{e}_i^2 = 0$.
</div>

<div class="fragment">
To represent Euclidean geometry in $n$ dimension, use $\mathbb{R}_{n,0,1}$:
    <ul>
        <li class="fragment">$n$ basis vector $\mathbf{e}_i$ with $\mathbf{e}_i^2 = 1$ to represent orthogonal (hyper)planes through the origin</li>
        <li class="fragment">$\mathbf{e}_0$ with $\mathbf{e}_0^2 = 0$ to represent the (hyper)plane at infinity.</li>
    </ul>
</div>
<div class="fragment">
Now the equation of a hyperplane in $\mathbb{R}^{n}$, given by $ a_1 x_1 + a_2 x_2 + \ldots + a_n x_n = -b$, is represented as a vector $a$ in the GA $\mathbb{R}_{n,0,1}$:
<div style="display: flex; gap: 1em; align-items: flex-start;">
  <div data-markdown style="flex: 1;">
$$ a = a_1 \mathbf{e}_1 + a_2 \mathbf{e}_2 + \ldots + a_n \mathbf{e}_n + b \mathbf{e}_0 \ .$$
  </div>
</div>
</div>

## Reflections all the way down

Geometry really is reflections all the way down! 

> Elements = Transformations

> **Conjugation** or **group action** is just composition of reflections! A $k$-reflection $A = a_1 \cdots a_k$ acts on on $\ell$-reflection $B = b_1 \cdots b_\ell$ as $$A[B] = (-1)^{k \ell} A B A^{-1}$$

> Geometric gauges allow you to reason about geometry


## Geometric Tolerance Analysis (GTA)

Geometric Tolerance Analysis is formulated in 3DPGA $\mathbb{R}_{3,0,1}$ but will also work as-is in 2DPGA.
- $\mathbb{R}_{3,0,1}$ has $3$ basis vectors $\mathbf{e}_{i}$ satisfying $\mathbf{e}_{i}^2=1$ to represent three orthogonal planes through the origin, and a single basis vector $\mathbf{e}_{0}$ satisfying $\mathbf{e}_{0}^2 = 0$ to represent the plane at infinity. 
- As a GA it is *graded*: scalars $s$, vectors <span style="color:#b4a0e8">$v = \sum_i v^i \mathbf{e}_{i}$</span>, bivectors <span style="color:#e88a8a">$B = \sum_{i < j} B^{ij} \mathbf{e}_{ij}$</span>, trivectors <span style="color:#e8a0c8">$t = \sum_{i < j < k} t^{ijk} \mathbf{e}_{ijk}$</span> and a quadvector <span style="color:#89d4e8">$q = q^{0123} \mathbf{e}_{0123}$</span>,  where <span style="color:#b4a0e8">$v^i$</span>, <span style="color:#e88a8a">$B^{ij}$</span>, <span style="color:#e8a0c8">$t^{ijk}$</span>, <span style="color:#89d4e8">$q^{0123}$</span> $\in \mathbb{R}$.

## Flat Features


a **plane** $p$ is characterized by a ~~normal~~ vector $\mathbf{n} = n_i \mathbf{e}_i$ and a distance $\delta$ (through/from the origin):
<div style="display: flex; gap: 1em; align-items: flex-start; align-items: center;">
  <div data-markdown style="flex: 1;">
$$ p = \mathbf{n} - \delta \mathbf{e}_0 \quad \leftrightarrow \quad \left(\begin{smallmatrix}n_1 \\ n_2 \\ n_3 \\ -\delta \end{smallmatrix}\right) $$
  </div>
  <div style="flex: 1;">
    <pre>
>>> alg = Algebra(3, 0, 1)
>>> p = alg.vector(e0='d', e1='n_1', e2='n_2', e3='n_3')
    </pre>
  </div>
</div>

A **point** $Q$ at location $\mathbf{q} = x \mathbf{e}_1 + y \mathbf{e}_2 + z \mathbf{e}_3$ is represented as a *pseudovector*:
<div style="display: flex; gap: 1em; align-items: flex-start; align-items: center;">
  <div data-markdown style="flex: 1;">
$$ Q = \mathbf{e}_{123} + \mathbf{q} \mathbf{e}_{0123} \quad \leftrightarrow \quad \left(\begin{smallmatrix}x \\ y \\ z \\ 1 \end{smallmatrix}\right) $$
  </div>
  <div style="flex: 1;">
    <pre>
>>> alg = Algebra(3, 0, 1)
>>> Q = alg.vector(e0=1, e1='x', e2='y', e3='z').dual()
    </pre>
  </div>
</div>
(Counter intuitive at first, but group theory forced us to this conclusion.)

A **line** $\ell$ is a *bivector* satisfying the Plücker condition $\ell \wedge \ell = 0$:
<div style="display: flex; gap: 1em; align-items: flex-start; align-items: center;">

  </div>
  <div style="flex: 1;">
    <pre>
>>> p1 ^ p2
>>> Q1 & Q2
    </pre>
  </div>

# The Two Norms

PGA is endowed with two norms: the **Euclidean norm** and the **Ideal norm**.
> Euclidean norm of a multivector $X$: $$\lVert X \rVert = \sqrt{X \widetilde{X}} \ .$$

Because $\mathbf{e}_{0}^2 = 0$, the ideal terms at infinity do not contribute to the norm, and hence it is also referred to as the Euclidean norm.
However, we are also interested in the norm of the ideal elements, which can be extracted using the ideal norm:
> Ideal norm of a multivector $X$: $$\lVert X \rVert_\infty = \lVert *X \rVert = \lVert X \vee *\mathbf{e}_{0} \rVert \ , $$
where $*X$ is the Hodge dual and $\vee$ is the regressive product.

### Measurement of Distance and Angle

<iframe src="https://enkimute.github.io/ganja.js/examples/coffeeshop.html#2GZf_mKS5q&fullscreen" width="100%" height="150px" frameBorder="0" alt="iframe"></iframe>

<center><i>Functional Element (FE) between two features $f_{i-1}$ and $f_{i}$: $ \boxed{F_i = f_{i-1} \widetilde{f_i} }$</i></center>

The distance and angle can be extracted using
$$
    d(F_i) = \begin{cases}
        \lVert \frac{\langle F_i \rangle_{k}}{\langle F_i \rangle_{k-2}} \rVert_\infty & \text{if } \lVert \langle F_i \rangle_{k-2} \rVert = \pm 1 \\
        0 & \text{otherwise}
    \end{cases} \ ,
    \qquad \qquad
    \theta(F_i) = \arctan \lVert \frac{\langle F_i \rangle_{\ell+2}}{\langle F_i \rangle_{\ell}} \rVert,
$$
$k$ is the *largest* integer for which $\langle F_i \rangle_k \neq 0$ and $\ell \geq 0$ is the *smallest* integer for which $\langle F_i \rangle_\ell \neq 0$.

> The measurements $d(F)$ and $\theta(F)$ are invariant under (special) Euclidean transformations $R \in \text{E}(3)$: $R[F] = R F \widetilde{R}$.

## Exponential and Logarithmic map

We have seen that rigid body transformations (rotations, translations, screws) are $2k$-reflections: $R = a b$ for rotations and translations, and $R = a b c d$ for screws.
These can be written in exponential form 

$$ \boxed{ R = e^{-\tfrac{1}{2} B} }$$

where

$$ B = \alpha \mathbf{e}_{23} + \beta \mathbf{e}_{31} + \gamma \mathbf{e}_{12} + u \mathbf{e}_{01} + v \mathbf{e}_{02} + w \mathbf{e}_{03} \quad \leftrightarrow \quad \begin{pmatrix}\alpha & \beta & \gamma & u & v & w \end{pmatrix}^T. $$

(The $-\tfrac{1}{2}$ is conventionally included such that for $B = \theta \mathbf{e}_{12}$, $X' = R X R^{-1}$ is a rotation over $\theta$.)

> The bivectors of 3DPGA are isomorphic to the torsors of UJTM and SDT.


## Infinitesimal transformations.

For small $B$, the rigid body transformation can be approximated by $R = 1 - \tfrac{1}{2} B$ and hence the transformation law becomes
$$ Y' = R Y R^{-1} \approx Y - \tfrac{1}{2} BY + \tfrac{1}{2}YB = Y + Y \times B $$
where $Y \times B = \tfrac{1}{2} [Y, B]$ is the *commutator product*.

> Since tolerances are by definition small, we will use this approximation in GTA when representing deviations from the nominal design.

The FE between any two normalized features $f_{i-1}$ and $f_i$ is $ \boxed{F_i = f_{i-1} \widetilde{f_i} }$
with $F_0 \coloneqq f_{n} \widetilde{f_0}$.

For a closed kinematic chain of FE's $\{ F_0, F_1, \ldots, F_n \}$ between $n+1$ features we have
$$ F_0 F_1 F_2 \cdots F_n = 1 \ . $$

Hence, if we take $F = F_0$ to be our Functional Requirement (FR), we find
$$ \boxed{ F_0 =  F_1 F_2 \cdots F_n } \ . $$

> The functional requirement is defined by the product of all FE's, **independent** of their types!

Within the FE $F_i = f_{i-1} \widetilde{f_i}$, the feature $f_{i-1}$ is seen as the reference, whereas $f_{i}$ is subject to change due to tolerances.
Since these tolerance zones are defined in the local reference frame of $f_i$, 
each tolerance zone $T_i$ needs to be placed in the global reference frame using a transformation $M_{i} \in \text{SE}(3)$:
$$ T = \sum_{i=1}^n M_i T_i \widetilde{M}_i \ .$$


## GTA Methodology: Perturbative GTA

The torsor encoding the perturbative effect of all tolerance zones is given by
$$ T = \sum_{i=1}^n M_i T_i \widetilde{M}_i \ ,$$
where $M_{i} \in \text{SE}(3)$ places $T_i$ in the correct position in the global reference frame.

Thus, the perturbed FR is given by
$$ F' = f_n' \widetilde{f_0} = (f_n + T \times f_n) \widetilde{f}_0 = F + (T \times f_n) \widetilde{f}_0 \ , $$
from which we can extract the new distance $d(F')$ and angle $\theta(F')$.

## GTA Methodology: Example

In the body frame of each feature we have the small displacement torsors
$$
\begin{aligned}
    T_{1} &= \alpha_1 \mathbf{e}_{23} + \beta_1 \mathbf{e}_{31} + w_1 \mathbf{e}_{03} \\
    T_{3} &= \alpha_2 \mathbf{e}_{23} + \gamma_2 \mathbf{e}_{12} + u_2 \mathbf{e}_{01} + w_2 \mathbf{e}_{03} \\
    T_{4} &= \alpha_3 \mathbf{e}_{23} + \gamma_3 \mathbf{e}_{12} + u_3 \mathbf{e}_{01} + w_3 \mathbf{e}_{03} \\
    T_{5} &= \alpha_4 \mathbf{e}_{23} + \gamma_4 \mathbf{e}_{12} + u_4 \mathbf{e}_{01} + w_4 \mathbf{e}_{03} \ .
\end{aligned}
$$
Then the other FE's are defined relative to the global reference frame by the motors
$$
\begin{aligned}
    M_1 &= 1 + \tfrac{1}{2} (110 \mathbf{e}_{02} + 55 \mathbf{e}_{03}) , \\
    M_3 &= M_4 = 1 + \tfrac{1}{2} (85 \mathbf{e}_{02}), \quad M_5 = 1. 
\end{aligned}
$$
By applying these motors to each respective tolerance zone $T_i$ we form the torsor that affects the FR:
$$
\begin{aligned}
    T &= M_1 T_{1} \widetilde{M}_1 + M_3 T_{3} \widetilde{M}_3 + M_4 T_{4} \widetilde{M}_4 + M_5 T_{5} \widetilde{M}_5 \\ 
    &= (u_2 + u_3 + u_4 + 55\beta_1 - 85\gamma_2 - 85\gamma_3) \mathbf{e}_{01} -55\alpha_1 \mathbf{e}_{02} \\
    &\quad + (w_1 + w_2 + w_3 + w_4 + 110\alpha_1 + 85\alpha_2 + 85\alpha_3) \mathbf{e}_{03} \\
    &\quad + (\gamma_2 + \gamma_3 + \gamma_4) \mathbf{e}_{12} + \beta_1 \mathbf{e}_{31} + (\alpha_1 + \alpha_2 + \alpha_3 + \alpha_4) \mathbf{e}_{23}
\end{aligned}
$$
We might be tempted to read of the $\mathbf{e}_{03}$ coefficient of this torsor and call it a day. However, we might not always be so lucky that our torsor aligns so nicely with the coordinate axes, so we should follow protocol to obtain a **coordinate independent** solution.

## GTA Methodology: Example

$$
\begin{aligned}
    T &= M_1 T_{1} \widetilde{M}_1 + M_3 T_{3} \widetilde{M}_3 + M_4 T_{4} \widetilde{M}_4 + M_5 T_{5} \widetilde{M}_5 \\ 
    &= (u_2 + u_3 + u_4 + 55\beta_1 - 85\gamma_2 - 85\gamma_3) \mathbf{e}_{01} -55\alpha_1 \mathbf{e}_{02} \\
    &\quad + (w_1 + w_2 + w_3 + w_4 + 110\alpha_1 + 85\alpha_2 + 85\alpha_3) \mathbf{e}_{03} \\
    &\quad + (\gamma_2 + \gamma_3 + \gamma_4) \mathbf{e}_{12} + \beta_1 \mathbf{e}_{31} + (\alpha_1 + \alpha_2 + \alpha_3 + \alpha_4) \mathbf{e}_{23}
\end{aligned}
$$
Instead, we should apply the displacement torsor to $f_5$ and recompute the FR:
$$
\begin{aligned}
    F' &= (f_5 + T \times f_5) \widetilde{f}_0 \\
    &= 55 \alpha_1 \mathbf{e}_{01} + (u_2 + u_3 + u_4 - 85 \gamma_2 - 85 \gamma_3 + 55 \beta_1) \mathbf{e}_{02} + \mathbf{e}_{12} \\
    &\; + (w_1 + w_2 + w_3 + w_4 + 110 \alpha_1 + 85 \alpha_2 + 85 \alpha_3 - 40) \mathbf{e}_{0123}  \ .
\end{aligned}
$$
Thus, the distance between $f_0$ and $f_5$ is given by $d(F') = \lVert \frac{\langle F' \rangle_4}{\langle F' \rangle_2} \rVert_\infty$: 
% Moreover, $\mathbf{e}_{0i}\mathbf{e}_{0123} = 0$, and so
$$
d(F') = \lvert w_1 + w_2 + w_3 + w_4 + 110 \alpha_1 + 85\alpha_2 + 85\alpha_3 - 40 \rvert \ .
$$
This is exactly the expected deviation around the nominal distance of $40$ due to the tolerance zones.

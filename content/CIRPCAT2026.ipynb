## An Algebra of Geometry

<div>
The rules I have explained are exactly the rules of Geometric Algebra (GA):
a Geometric (Clifford) Algebra $\mathbb{R}_{pqr}$ of $d$ = $p+q+r$ dimensions has $d$ anti-commuting ($\mathbf{e}_i \mathbf{e}_j=-\mathbf{e}_j \mathbf{e}_i$) basis vectors $\mathbf{e}_i$, of which $p$ are positive basis vectors $\mathbf{e}_i^2 = 1$, $q$ are negative basis vectors $\mathbf{e}_i^2 = -1$, and $r$ are null basis vectors $\mathbf{e}_i^2 = 0$.
</div>

<div class="fragment">
To represent Euclidean geometry in $n$ dimension, use $\mathbb{R}_{n,0,1}$:
    <ul>
        <li class="fragment">$n$ basis vector $\mathbf{e}_i$ with $\mathbf{e}_i^2 = 1$ to represent orthogonal (hyper)planes through the origin</li>
        <li class="fragment">$\mathbf{e}_0$ with $\mathbf{e}_0^2 = 0$ to represent the (hyper)plane at infinity.</li>
    </ul>
</div>
<div class="fragment">
Now the equation of a hyperplane in $\mathbb{R}^{n}$, given by $ a_1 x_1 + a_2 x_2 + \ldots + a_n x_n = -b$, is represented as a vector $a$ in the GA $\mathbb{R}_{n,0,1}$:
<div style="display: flex; gap: 1em; align-items: flex-start;">
  <div data-markdown style="flex: 1;">
$$ a = a_1 \mathbf{e}_1 + a_2 \mathbf{e}_2 + \ldots + a_n \mathbf{e}_n + b \mathbf{e}_0 \ .$$
  </div>
</div>
</div>

## Reflections all the way down

Geometry really is reflections all the way down! 

> Elements = Transformations

> **Conjugation** or **group action** is just composition of reflections! A $k$-reflection $A = a_1 \cdots a_k$ acts on on $\ell$-reflection $B = b_1 \cdots b_\ell$ as $$A[B] = (-1)^{k \ell} A B A^{-1}$$

> Geometric gauges allow you to reason about geometry

### Measurement of Distance and Angle

<iframe src="https://enkimute.github.io/ganja.js/examples/coffeeshop.html#2GZf_mKS5q&fullscreen" width="100%" height="150px" frameBorder="0" alt="iframe"></iframe>

<center><i>Functional Element (FE) between two features $f_{i-1}$ and $f_{i}$: $ \boxed{F_i = f_{i-1} f_i^{-1} }$</i></center>

The distance and angle between $f_{i-1}$ and $f_{i}$ can be extracted using (*valid only for $\leq 4$ dimensions)
$$
    d(F_i) = \begin{cases}
        \lVert \frac{\langle F_i \rangle_{k}}{\langle F_i \rangle_{k-2}} \rVert_\infty & \text{if } \lVert \langle F_i \rangle_{k-2} \rVert = \pm 1 \\
        0 & \text{otherwise}
    \end{cases} \ ,
    \qquad \qquad
    \theta(F_i) = \arctan \lVert \frac{\langle F_i \rangle_{\ell+2}}{\langle F_i \rangle_{\ell}} \rVert,
$$
<center><small>
$k$ is the <i>largest</i> integer for which $\langle F_i \rangle_k \neq 0$ and $\ell \geq 0$ is the <i>smallest</i> integer for which $\langle F_i \rangle_\ell \neq 0$.
</small></center>

<div class="fragment" data-markdown>

> The measurements $d(F)$ and $\theta(F)$ are coordinate-independent and do not involve coordinate projections.

</div>
<div class="fragment" data-markdown>

> The measurements $d(F)$ and $\theta(F)$ are invariant under (special) Euclidean transformations $R \in \text{E}(3)$: $R[F] = R F R^{-1}$.

</div>

with $F_0 \coloneqq f_{n} f_0^{-1}$.

## GTA Methodology: Perturbative GTA

The torsor encoding the perturbative effect of all tolerance zones is given by
$$ T = \sum_{i=1}^n M_i T_i M_i^{-1} \ ,$$
where $M_{i} \in \text{SE}(3)$ places $T_i$ in the correct position in the global reference frame.

<div class="fragment" data-markdown>

This generates a pertubation $e^{-\tfrac{1}{2} T} \approx 1 - \tfrac{1}{2} T$ on $f_n$: 
    $$ f_n' = e^{-\tfrac{1}{2} T} f_n e^{\tfrac{1}{2} T} \approx f_n + \tfrac{1}{2} [f_n, T ] \ . $$
</div>

<div class="fragment" data-markdown>

Thus, the **perturbed FR** $F'$ is given by
$$ \boxed{F' = f_n' f_0^{-1} \approx (f_n + T \times f_n) f_0^{-1} = F + \tfrac{1}{2} [f_n, T ] f_0^{-1}} \ , $$
from which we can extract the new distance $d(F')$ and angle $\theta(F')$.

</div>

<div style="display: flex; gap: 1em; align-items: flex-start; align-items: center;">

## Exponential and Logarithmic map

Rigid body transformations (rotations, translations, screws) can be written in exponential form 

$$ \boxed{ R = e^{-\tfrac{1}{2} B} }$$
$$ B = \alpha \mathbf{e}_{23} + \beta \mathbf{e}_{31} + \gamma \mathbf{e}_{12} + u \mathbf{e}_{01} + v \mathbf{e}_{02} + w \mathbf{e}_{03} \quad \leftrightarrow \quad \begin{pmatrix}\alpha & \beta & \gamma & u & v & w \end{pmatrix}^T. $$

(The $-\tfrac{1}{2}$ is conventionally included such that for $B = \theta \mathbf{e}_{12}$, $X' = R X R^{-1}$ is a rotation over $\theta$.)

> Bivectors in 3DPGA are isomorphic to the torsors of UJTM and SDT.

<div class="fragment" data-markdown>

**Infinitesimal transformations**

For small $B$, the rigid body transformation can be approximated by $R = 1 - \tfrac{1}{2} B$ and hence the
$$ Y' = R Y R^{-1} \approx Y - \tfrac{1}{2} BY + \tfrac{1}{2}YB = Y + Y \times B $$
where $Y \times B = \tfrac{1}{2} [Y, B]$ is the *commutator product*.

> Since tolerances are by definition small, we use this approximation when representing deviations from the nominal design.

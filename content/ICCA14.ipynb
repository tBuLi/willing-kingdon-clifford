# Invariant Decomposition

**Cartan-Dieudonné**: 
> Every orthogonal transformation in an $n$-dimensional embedding space is composed from at most $n$ reflections.

**Invariant Decomposition**:

> A product $R=r_1 r_2 \cdots r_n \in \text{Pin}(p,q,r)$ of $n$ reflections $r_i$ can be decomposed into exactly $\lceil \frac n 2 \rceil$ simple commuting factors. These are $k = \lfloor \frac n 2 \rfloor$ bireflections, and, for odd $n$, one extra reflection.

</div>

## Bivector Split
**Q**: Given $R = \prod_{i=1}^k R_i = \prod_{i=1}^k u_i v_i \in \text{Spin}(p,q,r)$, find the **commuting** bireflections $R_i = u_i v_i$.
<div class="fragment" data-fragment-index="0">

**A**: Since $R = \exp(B), \implies$
Given a bivector $B$ how do we split it into mutually commuting $B_i$?

</div>
<div class="fragment" data-fragment-index="1">

Assume the solution $ B = B_1 + B_2 + \ldots + B_k $ exists. 
Then the $B_i$ must be roots of
$$ (B_1 - B_i) (B_2 - B_i) \cdots (B_k - B_i) = 0. $$

</div>
<div class="fragment" data-fragment-index="2">

Solve for $B_i$:

</div>
<div class="r-stack">
<div class="fragment current-visible" data-fragment-index="2">

$$ B_i = \frac{B + \tfrac{1}{3! \lambda_i} B \wedge B \wedge B + \ldots}{1 + \tfrac{1}{2! \lambda_i} B \wedge B  + \ldots}, \quad \text{where} \quad B_i^2 = \lambda_i \in \mathbb{R} $$

</div>
<div class="fragment" data-fragment-index="3">

$$ B_i = \frac{W_1 + \tfrac{1}{\lambda_i} W_3 + \tfrac{1}{\lambda_i^2} W_5 + \ldots}{W_0 + \tfrac{1}{\lambda_i} W_2  + \tfrac{1}{\lambda_i^2} W_4 + \ldots}, \qquad  W_j = \tfrac{1}{j!} \langle B^j \rangle_{2j}$$

</div>
</div>
<div class="fragment" data-fragment-index="4">

The $\lambda_i = B_i^2$ are the roots of $0 = \sum_{j=0}^k \langle W_j^2 \rangle_{0} (- \lambda_i)^{k-j} $

</div>

## Isoclinic rotations

<b>Q</b>: What happens for degenerate eigenvalues? 
Take e.g. the **spine** in $\mathbb{R}_{2k}$

$$ \Sigma_k \coloneqq B_1 + \ldots + B_k = \mathbf{e}_1 \bar{\mathbf{e}}_1 + \ldots + \mathbf{e}_k \bar{\mathbf{e}}_k $$

with $\lambda := B_1^2 = \ldots = B_k^2 = -1$. (w.l.o.g., the $B_i$ form commuting complex units.)

<b>A</b>: The bivector split is no longer unique. Simple example in $m = 2k$:

$$ \Sigma_2 = B_1 + B_2 = B_1 (1 - \underbrace{B_1 B_2}_{\tfrac{1}{2} \Sigma_2 \wedge \Sigma_2}) \implies B_1 = \Sigma_2 (1 - \tfrac{1}{2} \Sigma_2 \wedge \Sigma_2)^{-1}$$

but $(1 - B_1 B_2)$ is singular when $\lambda_1 = \lambda_2 = -1$, and so there is no longer a unique solution.

> Claim: the **spine** $\Sigma_k$ defines **spinors**

## Intermezzo: Traditional Spinors (in GA language)

<b>In $m=4$ ($k=2$) dimensions</b>,
$$ B_j^2 = -1 = i^2 \implies (B_j - i )(B_j + i ) = 0 \implies B_j (B_j + i ) = i (B_j + i )$$

What this tell us is that on the *idempotent* 
$$f_j \coloneqq \tfrac{1}{2}(B_j + i ) \qquad (f_j = v_{+j}v_{-j}) $$

- $B_j$ behaves like $i$ since $B_j f = i f$.
- $f_j$ cannot be invertable as this would imply $B_j = i$.

<b>In $m=2k$ dimensions</b>, define $\boxed{f \coloneqq \prod_{j=1}^k f_j}$.
On the idempotent $f$ all "complex units" are the same:
$$\boxed{B_1 f = \ldots = B_k f = i f} $$

**Spinor space** $\mathcal{S}_\mathbb{C} = \mathbb{C}_{2k} f, \qquad \dim(\mathcal{S}_\mathbb{C}) = 2^{k}$ complex dof.

## Intermezzo: Traditional Spinors (in GA language)

$$ f \coloneqq \prod_{j=1}^k f_j = \tfrac{1}{2^k}\prod_{j=1}^k (B_j + i ) , \qquad f^2 = f, \qquad B_j f = i f $$


$$
\small{\begin{aligned}
& & \textbf{Real part} & & \textbf{Imaginary part} \\
k=2: \quad & f = & -(1 - B_1 B_2) & + & i(B_1 + B_2) \\
k=3: \quad & f = & -(B_1 + B_2 + B_2 + B_1B_2B_3) & + & -i(1 - B_1 B_2 - B_1 B_3 - B_2 B_3) \\
& & \vdots & & \vdots
\end{aligned}}$$

<div class="fragment">

$$
\small{\begin{aligned}
k=2: \quad \forall j & & B_j (1 - B_1 B_2) &= B_1 + B_2 \\
k=3: \quad \forall j & & -B_j (B_1 + B_2 + B_2 - B_1B_2B_3) &= (1 - B_1 B_2 - B_1 B_3 - B_2 B_3) \\
& & &\vdots
\end{aligned}}$$
> In a sense, the **real** or **imaginary** part is enough here!

</div>

<div class="fragment">

> Claim: the *real* spinor space $\mathcal{S}_\mathbb{R} = \mathbb{R}_k \Box$ defined by the **forgetful idempotent**
> $$\Box = \tfrac{1}{2^k}(1 - B_1B_2) (1 - B_2B_3) \cdots (1 - B_{k-1}B_k) = c_\wedge(i \Sigma_k)$$
> and carries the same information as $\mathcal{S}_\mathbb{C}$.

</div>

## Box Technology: Dimensional Counting

Let us drive this points home by doing a dimensional counting for both $\mathcal{S}_\mathbb{C} = \mathbb{C}_{m} f$ and $\mathcal{S}_\mathbb{R} = \mathbb{R}_{m} \Box$:

<div class="fragment">
Recall that $\Box$ equates $2^{k-1}$ elements of the algebra, while $f$ equates $2^k$:
$$\begin{aligned}
\Box &= -B_i B_j \Box = B_i B_j B_k B_l \Box = \ldots \rightarrow & 1 + \tbinom{k}{2} + \tbinom{k}{4} + \ldots &= 2^{k-1} \\
f &= -B_i B_j f = -i B_j f = \ldots \rightarrow & 1 + \tbinom{k+1}{2} + \tbinom{k+1}{4} + \ldots &= 2^{k} \\
\end{aligned}$$
</div>

<div class="r-hstack fragment" style="gap: 2em; align-items: start;">
<div>

**Spine-ors**
- $\dim(\mathbb{R}_{2k}) = 2^{2k} \text{ real}$
- $\dim(\mathbb{R}_{2k} \Box) = \color{#ffb86c}2^{k+1} \text{ real}$

</div>
<div>

**Classical Approach**
- $\dim(\mathbb{C}_{2k}) = 2^{2k} \text{ complex} = 2^{2k+1} \text{ real}$
- $\dim(\mathbb{C}_{2k} f) = 2^k \text{ complex} = \color{#ffb86c}2^{k+1} \text{ real}$

</div>
</div>

<div class="fragment">

> Conclusion: $\dim(\mathbb{C}_{m} f) = \dim(\mathbb{R}_{m} \Box)$!
>
> $i = \sqrt{-1}$ does not introduce any new information to the spinor space.

</div>

<div class="fragment">

> Nuance: $i$ does introduce new information to the GA $\mathbb{R}_m \to \mathbb{C}_m$, but since one studies spinors as representations of the **real Lie groups** $\text{Spin}(m)$ and $\text{SU}(n)$, those extra d.o.f. are redundant.

</div>

## Box Technology: Remarks on "Reality of Spinors"

- Realising spinors via the box (**spine-ors**) provides a way to make spinors in a **real GA**
- However, we prefer *not* to call them "real spinors", because that term is usually reserved for (real) Majorana spinors.
    - Majorana spinors can *also* be made in the **spine-or** setting, as we will see if I have time.

## Graded Structure of $\mathcal{S}$

General $m=2k$:
- $\mathcal{S} = \mathbb{R}_{2k} \Box = \mathbb{R}_{k+1} \Box \cong (\mathbb{R}_{k} \otimes \mathbb{C}) \Box$ .
- $\mathcal{S}^j \coloneqq \{ X \Box + Y \Box B_1 : X, Y \in \mathbb{R}_k^j \}$
- $\mathcal{S} = \mathcal{S}^+ \oplus \mathcal{S}^-$ decomposes into two $\text{Spin}(2k)$-irreps (Weyl spinors).
- The $\mathcal{S}^j$ are $\text{SU}(k)$-irreps
  $$\begin{aligned} 
  \text{SU}(k) &= \{ s \in \text{Spin}(2k) : s \Box = \Box \widetilde{s} = \Box \} \\
  \mathfrak{su}(k) &= \{ B \in \mathbb{R}_{2k}^2 : B \Box = 0 \} 
  \end{aligned}$$


Taking $m=4$ as an example, $\mathbb{R}_4 = \text{Alg}(\mathbf{e}_1, \bar{\mathbf{e}}_1, \mathbf{e}_2, \bar{\mathbf{e}}_2)$
- $\mathcal{S} = \mathbb{R}_4 \Box = \mathbb{R}_3 \Box $ where $\mathbb{R}_3 = \text{Alg}(\mathbf{e}_1, \bar{\mathbf{e}}_1, \mathbf{e}_2)$.
- $\mathbb{R}_3$ is graded: $\mathbb{R}_3 = \bigoplus_j \mathbb{R}_3^j$.
    - So it feels natural to take $\mathcal{S}^j = \mathbb{R}_3^j \Box$. **But** this is *not* the best idea because this ignores the natural identification of $B_1 = \mathbf{e}_1 \bar{\mathbf{e}}_1$ with complex numbers.
- We define instead $\mathcal{S}^j \coloneqq \{ X \Box + Y \Box B_1 : X, Y \in \mathbb{R}_2^j \}$, s.t. $\mathcal{S} = \underbrace{\mathcal{S}^0}_{\mathcal{S}^+} \oplus \underbrace{\mathcal{S}^1}_{\mathcal{S}^-} \oplus \underbrace{\mathcal{S}^2}_{\mathcal{S}^+}$


## SU(k)-irreps: a leptonic example
<sup><sub>Roelfs & Eelbode, Lepton Triptych I: Geometric Foundations of Electroweak Symmetry in the Real Clifford Algebra $Cl_4(\mathbb{R})$</sub></sup>

For $m=4$ (i.e. $k=2$), we have
- $\Sigma_2 = B_1 + B_2 = \mathbf{e}_1\bar{\mathbf{e}}_1 + \mathbf{e}_2\bar{\mathbf{e}}_2$
- $\Box = \tfrac{1}{2} (1 - B_1 B_2)$
- $B_1 \Box = B_2 \Box \implies (B_1 - B_2) \Box = 0 \implies (1 + B_1 B_2) \Box = 0 $
- $(1 + B_1 B_2) \Box = 0 $ leads to three annihilating bivectors:
$$\begin{aligned}
\rho_1 &= \tfrac{1}{2}\mathbf{e}_1\mathbf{e}_2 (1 + B_1 B_2) &\leftrightarrow\quad i\sigma_1 &= \bigl(\begin{smallmatrix}0&i\\i&0\end{smallmatrix}\bigr) \\
\rho_2 &= \tfrac{1}{2}\mathbf{e}_1\bar{\mathbf{e}}_2 (1 + B_1 B_2) &\leftrightarrow\quad i\sigma_2 &= \bigl(\begin{smallmatrix}0&1\\-1&0\end{smallmatrix}\bigr) \\
\rho_3 &= \tfrac{1}{2}\mathbf{e}_1\bar{\mathbf{e}}_1 (1 + B_1 B_2) &\leftrightarrow\quad i\sigma_3 &= \bigl(\begin{smallmatrix}i&0\\0&-i\end{smallmatrix}\bigr)

## Spine-ors and Cayley-Dickson $\tiny{\mathbb{R} \to \mathbb{C} \to \color{#ffb86c}\mathbb{H} \color{#bd93f9}\to \mathbb{O} \to \mathbb{S} \to \ldots}$

For $m=4$ with basis $\{ \mathbf{e}_1, \bar{\mathbf{e}}_1,  \mathbf{e}_2, \bar{\mathbf{e}}_2 \}$, we have
$$\Sigma_2 = B_1 + B_2 = \mathbf{e}_1\bar{\mathbf{e}}_1 + \mathbf{e}_2\bar{\mathbf{e}}_2 \qquad 
\Box = \tfrac{1}{2} (1 - B_1 B_2) = \tfrac{1}{2} (1 - \mathbf{e}_1\bar{\mathbf{e}}_1 \mathbf{e}_2\bar{\mathbf{e}}_2) $$


> Claim: we can recover $\mathbb{H}$ from $\mathbf{e}_1\mathbf{e}_2, \mathbf{e}_1\bar{\mathbf{e}}_1, \mathbf{e}_1\bar{\mathbf{e}}_2$.
- Seems on odd claim, since e.g. $(\mathbf{e}_1\mathbf{e}_2)(\mathbf{e}_1\bar{\mathbf{e}}_1) = \bar{\mathbf{e}}_1\mathbf{e}_2 $ is not on the list!
- However,
  $$ \bar{\mathbf{e}}_1\mathbf{e}_2 \Box =  \bar{\mathbf{e}}_1\mathbf{e}_2 + (\bar{\mathbf{e}}_1\mathbf{e}_2) B_1 B_2 = \bar{\mathbf{e}}_1\mathbf{e}_2 + \mathbf{e}_1 \bar{\mathbf{e}}_2 = \mathbf{e}_1 \bar{\mathbf{e}}_2 \Box$$
- Hence, quaternion product defined via $\Box$ using $\mathbf{e}_1 \mathbb{R}^4 \Box$.

## Spine-ors and Cayley-Dickson $\tiny{\mathbb{R} \to \mathbb{C} \to \mathbb{H} \to \color{#ffb86c}\mathbb{O} \color{#bd93f9}\to \mathbb{S} \to \ldots}$

Algebra structure on $\mathbf{e}_0 \mathbb{R}^m$ with $m = 2^n$ ($n \geq 2$):
  $$ \mathcal{A} = \text{span}(\{ 1, \underbrace{\mathbf{e}_0 \mathbf{e}_1, \mathbf{e}_0 \mathbf{e}_2, \ldots, \mathbf{e}_0 \mathbf{e}_{m-1}}_\text{``complex units''} \})$$
Let us illustrate the next case: *octonions*
- Make **spine** $\Sigma_4 = B_1 + B_2 + B_3 + B_4 = \mathbf{e}_0 {\mathbf{e}}_1 + \ldots + \mathbf{e}_6 \mathbf{e}_7$
- Make **spinor space** $\mathcal{S}_8 = \mathbb{R}_8 \Box_4 \cong \mathbb{R}_4 \Box_4 \mathbb{C}$
- This means we still have an $\mathbb{R}_4$ left in the spinor space $\mathcal{S}_8$!
- So repeat; make **spine** $\Sigma_2 = b_1 + b_2 = \mathbf{e}_0 \mathbf{e}_2 + \mathbf{e}_4 \mathbf{e}_6 \in \mathbb{R}_4^2$

- This means we still have an $\mathbb{R}_2$ left in the spinor space $\mathcal{S}_4$. Trivial, since $\Sigma_1 = \mathbf{e}_0 \mathbf{e}_4$ and $\Box_1 = 1$.

> $$ \Box_\mathbb{O} = \Box_1 \Box_2 \Box_4 = \Box_2 \Box_4 $$

## Spine-ors and Cayley-Dickson $\tiny{\mathbb{R} \to \mathbb{C} \to \mathbb{H} \to \color{#ffb86c}\mathbb{O} \color{#bd93f9}\to \mathbb{S} \to \ldots}$
### Octonionic multiplication

- $\omega \in \mathbb{O}$ identified with $\mathbf{e}_0 v \in \mathbf{e}_0 \mathbb{R}^8$
- $\omega =\mathbf{e}_0 (c_0 \mathbf{e}_0 + \ldots + c_7 \mathbf{e}_7)$
- $\mathbb{O} = \text{span}(\{ 1, \underbrace{\mathbf{e}_0 \mathbf{e}_1, \mathbf{e}_0 \mathbf{e}_2, \ldots, \mathbf{e}_0 \mathbf{e}_7}_\text{``complex units''} \})$
- $ \Box_\mathbb{O} = \Box_2 \Box_4 $

> Lemma:
> - Take any $\beta_1, \beta_2 \in \{ \mathbf{e}_0 \mathbf{e}_1, \mathbf{e}_0 \mathbf{e}_2, \ldots, \mathbf{e}_0 \mathbf{e}_7 \}$
> - $\langle \beta_1 \beta_2 \Box_\mathbb{O} \rangle_2$ is a spine in $\mathbb{R}_8^2$, i.e. $\langle \beta_1 \beta_2 \Box_\mathbb{O} \rangle_2 = \beta_1 \beta_2 + B_1' + B_2' + B_3' $
> - One of these $B_j'$ contains $\mathbf{e}_0$ (w.l.o.g. $B_1'$):
>   $\langle \beta_1 \beta_2 \Box_\mathbb{O} \rangle_2 = \beta_1 \beta_2 \pm \mathbf{e}_0 \mathbf{e}_r + B_2' + B_3' $
> - $\beta_1 * \beta_2 \coloneqq \pm \mathbf{e}_0 \mathbf{e}_r$ defines the octonionic product

## Bivector Split

Starting from the eigen-equation for the bivector $B$, we can extend this to all $W_j = \tfrac{1}{j!} \langle B^j \rangle_{2j}$:
$$ B \cdot v_\mu = \mu v_\mu \implies W_j \cdot v_\mu = \mu \, W_{j-1} \wedge v_\mu $$

<div class="fragment">
If we define the outer exponential 
$$ \Lambda^B := \sum_{j=0}^k W_j = 1 + B + \tfrac{1}{2!} B \wedge B + \ldots, $$
then can start from $W_k v_\mu = W_k \cdot v_\mu + \cancel{W_k \wedge v_\mu}$ and apply the lemma recursivelly, to find
$$ B \cdot v_\mu = \mu v_\mu \implies \Lambda^{-B / \mu} v_\mu = 0 $$
</div>
<span class="fragment">
<b>What is the payoff?</b> Notice that the uninvertable inner product has been replaced by the <i>invertable</i> geometric product,
meaning that we only have non-trivial solutions when $\Lambda^{B / \mu} \Lambda^{- B / \mu} = 0$!
</span>
<span class="fragment">
The bivector split can then be shown to be equivalent to 
    $$ b_i = \frac{W_1 + \tfrac{1}{\lambda_i} W_3 + \tfrac{1}{\lambda_i^2} W_5 + \ldots}{W_0 + \tfrac{1}{\lambda_i} W_2  + \tfrac{1}{\lambda_i^2} W_4 + \ldots} \qquad \leftrightarrow \qquad b_i = \mu_i {\text{t}}_\wedge(B / \mu_i) = \mu_i \frac{\Lambda^{B / \mu_i} - \Lambda^{-B / \mu_i}}{\Lambda^{B / \mu_i} + \Lambda^{-B / \mu_i}} $$
    $$ \sum_{j=0}^k \langle W_j^2 \rangle_{0} (- \lambda_i)^{k-j} = 0 \qquad \leftrightarrow \qquad \Lambda^{B / \mu_i} \Lambda^{- B / \mu_i} = 0 $$
</span>

## Charge Conjugation & Majorana Spinors

Charge conjugation transforms particles into anti-particels and vice versa.

The charge conjugation operator $\mathcal{C}$ satisfies
- anti-linearity: $ \mathcal{C}(\lambda \Psi) = \bar{\lambda} \mathcal{C}(\Psi)$ where $\lambda \in \mathbb{C}$, $\forall \Psi \in \mathcal{S}$
- (anti-)idempotent: $\mathcal{C}^2 = \pm 1$
- transforms covariantly: $\mathcal{C}(\Lambda \Psi) = \Lambda\mathcal{C}( \Psi) \quad \forall \Lambda \in \text{Spin}(1,3)$


<div class="r-hstack" style="gap: 2em; align-items: start;">

<div>

**Classical Approach**

- $\mathcal{C}(\Psi) = C \Psi^*$

</div>

<div>

**Spine-ors**

- $\mathcal{C}(\Psi) = \Psi C$

</div>

</div>

> Claim: $C = \mathbf{e}_1 \cdots \mathbf{e}_k $. (As seen before for leptons and quarks.)

Since $C \Sigma_k = -\Sigma_k C$ the geometry of charge conjugation is evidently to flip the sign of each $B_j \to - B_j$, which also explains why the classical approach requires additional complex conjugation $i \to -i$.

## Box Technology

1. **forgetful idempotent** constructed via the "round table principle":
   $$\Box = \tfrac{1}{2^k}(1 - B_1B_2) (1 - B_2B_3) \cdots (1 - B_{k-1}B_k)$$
2. Connection with the spine via the outer cosine:   $\Box = \tfrac{1}{2^k} c_\wedge(i \Sigma)$
3. Subtle differences between $f$ and $\Box$:
    - We are all eating the same thing, called $i$:   $B_j f = i f \quad (\forall j)$
    - We are all eating the same thing:       $B_i \Box = B_j\Box \quad (\forall i,j)$

> (Bold) claim: $i = \sqrt{-1}$ is not required; spinors are not fundamentally complex. $\Box$ does the heavy lifting in both constructions.

## Formulating (a realization of) Geometry

The equation of a hyperplane in $\mathbb{R}^{p,q}$ 
    $$ a_1 x_1 + a_2 x_2 + \ldots + a_n x_n + b = 0$$
with $n = p + q$
can be represented as a vector $v$ in a Clifford algebra of dimension $d=n+1$:
$$ v = a_1 \mathbf{e}_1 + a_2 \mathbf{e}_2 + \ldots + a_n \mathbf{e}_n + b \mathbf{e}_0,$$
where $\mathbf{e}_i$ are the basis vectors of the algebra.

<span class="fragment">
Depending on the sign of $\mathbf{e}_0$ we get different geometries:

# Invariant Decomposition

**Cartan-Dieudonné**: 
> Every orthogonal transformation in an $n$-dimensional embedding space is composed from at most $n$ reflections.

**Invariant Decomposition**:

> A product $U=u_1 u_2 \cdots u_n \in \text{Pin}(p,q,r)$ of $n$ reflections $u_i$ can be decomposed into exactly $\lceil \frac n 2 \rceil$ simple commuting factors. These are $k = \lfloor \frac n 2 \rfloor$ bireflections, and, for odd $n$, one extra reflection.


## Invariant Decomposition of $U \in \text{Pin}(p,q,r)$

- **odd $n$**: $U = r R$, where $r$ is a single reflection and $R$ is a $2k$-reflection.
  $$r = \langle U \rangle_1, \qquad R = r^{-1} U = U r^{-1}, \qquad r \times R = \tfrac{1}{2} [r, R] = 0 $$
- **even $n$**: there exists a factorization of a $2k$-reflection $R$ into $k$ commuting bireflections $R_i$ such that
  $$R = \prod_{i=1}^k R_i = \prod_{i=1}^k u_i v_i$$

Every (simple) bireflection is of the form $R_i = \alpha_i + \beta_i B_i$, where $\alpha_i, \beta_i \in \mathbb{R}$ and $B_i^2 \in \mathbb{R}$. Hence this results in a rich variaty of multi-complex-like numbers!

## Bivector split

For a given $2k$-reflection $R = \prod_{i=1}^k R_i = \prod_{i=1}^k u_i v_i$, how do we actually find the **commuting** bireflections $R_i = u_i v_i$?

Assuming the factorization exists, we know that each bireflection is of the form 
$$R_i = \alpha_i + \beta_i B_i, $$
with $\alpha_i, \beta_i \in \mathbb{R}$ and $B_i$ a *simple* bivector: $B_i^2 = \lambda_i \in \mathbb{R}$.

So the bivector part of $R$ is a linear combination of the $B_i$:
$$ \langle R \rangle_2 = \sum_{i=1}^k (\prod_{j \neq i} \alpha_j) \beta_i B_i $$

So the **invariant factorization** reduces to the **bivector split**:
$$ B = B_1 + B_2 + \ldots + B_k $$

## Bivector Split

*Question*: Given a bivector $B$, how do we split it into mutually commuting $b_i$?

*Answer*: assume the solution $$ B = B_1 + B_2 + \ldots + B_k $$ exists. 
Then the $B_i$ must be roots of
$$ (B_1 - B_i) (B_2 - B_i) \cdots (B_k - B_i) = 0. $$

## Bivector Split

Starting from the eigen-equation for the bivector $B$, we can extend this to all $W_j = \tfrac{1}{j!} \langle B^j \rangle_{2j}$:
$$ B \cdot v_\mu = \mu v_\mu \implies W_j \cdot v_\mu = \mu \, W_{j-1} \wedge v_\mu $$

<div class="fragment">
If we define the outer exponential 
$$ \Lambda^B := \sum_{j=0}^k W_j = 1 + B + \tfrac{1}{2!} B \wedge B + \ldots, $$
then can start from $W_k v_\mu = W_k \cdot v_\mu + \cancel{W_k \wedge v_\mu}$ and apply the lemma recursivelly, to find
$$ B \cdot v_\mu = \mu v_\mu \implies \Lambda^{-B / \mu} v_\mu = 0 $$
</div>
<span class="fragment">
<b>What is the payoff?</b> Notice that the uninvertable inner product has been replaced by the <i>invertable</i> geometric product,
meaning that we only have non-trivial solutions when $\Lambda^{B / \mu} \Lambda^{- B / \mu} = 0$!
</span>
<span class="fragment">
The bivector split can then be shown to be equivalent to 
    $$ b_i = \frac{W_1 + \tfrac{1}{\lambda_i} W_3 + \tfrac{1}{\lambda_i^2} W_5 + \ldots}{W_0 + \tfrac{1}{\lambda_i} W_2  + \tfrac{1}{\lambda_i^2} W_4 + \ldots} \qquad \leftrightarrow \qquad b_i = \mu_i {\text{t}}_\wedge(B / \mu_i) = \mu_i \frac{\Lambda^{B / \mu_i} - \Lambda^{-B / \mu_i}}{\Lambda^{B / \mu_i} + \Lambda^{-B / \mu_i}} $$
    $$ \sum_{j=0}^k \langle W_j^2 \rangle_{0} (- \lambda_i)^{k-j} = 0 \qquad \leftrightarrow \qquad \Lambda^{B / \mu_i} \Lambda^{- B / \mu_i} = 0 $$
</span>

## Isoclinic rotations

<b>Q</b>: What happens for degenerate eigenvalues? 
Take e.g. the **spine** 

$$ \Sigma_k \coloneqq B_1 + B_2 + \ldots + B_k $$

with $\lambda := B_1^2 = \ldots = B_k^2 = -1$. (Suitable normalization s.t. the $B_i$ form commuting complex units.)

<b>A</b>: The bivector split is no longer unique. Simple example in $m = 2k$:

$$ \Sigma_2 = B_1 + B_2 = B_1 (1 - \underbrace{B_1 B_2}_{\tfrac{1}{2} \Sigma_2 \wedge \Sigma_2}) \implies B_1 = \Sigma_2 (1 - \tfrac{1}{2} \Sigma_2 \wedge \Sigma_2)^{-1}$$

but $(1 - B_1 B_2)$ is singular when $\lambda_1 = \lambda_2 = -1$, and so there is no longer a unique solution.

> Claim: the **spine** $\Sigma_k$ defines **spinors**

## Intermezzo: Traditional Spinors (in GA language)

<b>In $m=2$ dimensions</b>,
$$ B_j^2 = -1 = i^2 \implies (B_j - i )(B_j + i ) = 0 \implies B_j (B_j + i ) = i (B_j + i )$$

What this tell us is that on the *idempotent* 
$$\boxplus_j \coloneqq \tfrac{1}{2}(B_j + i ) \qquad (\boxplus_j = v_{+j}v_{-j}) $$

- $B_j$ behaves like $i$ since $B_j \boxplus = i \boxplus$.
- $\boxplus_j$ cannot be invertable as this would imply $B_j = i$.

<b>In $m=2k$ dimensions</b>, define $\boxed{\boxplus \coloneqq \prod_{j=1}^k \boxplus_j}$.
On the idempotent $\boxplus$ all "complex units" are the same:
$$\boxed{B_1 \boxplus = \ldots = B_k \boxplus = i \boxplus} $$

The spinor space is given by $\mathcal{S}_\mathbb{C} = \mathbb{C}_k \boxplus$ and $\dim(\mathcal{S}_\mathbb{C}) = 2^{k}$ complex dof.

## Box Technology

1. Constructed via the "round table principle":
   $$\Box = \tfrac{1}{2^k}(1 - B_1B_2) (1 - B_2B_3) \cdots (1 - B_{k-1}B_k)$$
2. Connection with the spine via the outer cosine:   $\Box = \tfrac{1}{2^k} \cos_\wedge(i \Sigma)$
3. Subtle differences between $\boxplus$ and $\Box$:
    - We are all eating the same thing, called $i$:   $B_j \boxplus = i \boxplus \quad (\forall j)$
    - We are all eating the same thing:       $B_i \Box = B_j\Box \quad (\forall i,j)$

> (Bold?) claim: $i = \sqrt{-1}$ is not really required here, so spinors are not fundamentally complex. $\Box$ does all the heavy lifting in both constructions.

## Box Technology: Dimensional Counting

Let us drive this points home by doing a dimensional counting for both $\mathcal{S}_\mathbb{C} = \mathbb{C}_{m} \boxplus$ and $\mathcal{S}_\mathbb{R} = \mathbb{R}_{m} \Box$:

Recall that $\Box$ equates $2^{k-1}$ elements of the algebra, while $\boxplus$ equates $2^k$:
$$\begin{aligned}
\Box &= -B_i B_j \Box = B_i B_j B_k B_l \Box = \ldots \rightarrow & 1 + \tbinom{k}{2} + \tbinom{k}{4} + \ldots &= 2^{k-1} \\
\boxplus &= -B_i B_j \boxplus = -i B_j \boxplus = \ldots \rightarrow & 1 + \tbinom{k+1}{2} + \tbinom{k+1}{4} + \ldots &= 2^{k} \\
\end{aligned}$$

## Graded Structure of $\mathcal{S}$

Taking $m=4$ as an example, 
- $\mathcal{S} = \mathbb{R}_4 \Box = \mathbb{R}_3 \Box $ where $\mathbb{R}_3 = \text{Alg}(\mathbf{e}_1, \bar{\mathbf{e}}_1, \mathbf{e}_2)$.
- $\mathbb{R}_3$ is graded: $\mathbb{R}_3 = \bigoplus_j \mathbb{R}_3^j$.
    - So it feels natural to take $\mathcal{S}^j = \mathbb{R}_3^j \Box$. **But** this is *not* the best idea because this ignores the natural identification of $B_1 = \mathbf{e}_1 \bar{\mathbf{e}}_1$ with complex numbers.
- We define instead $\mathcal{S}^j \coloneqq \{ X \Box + Y \Box B_1 : X, Y \in \mathbb{R}_2^j \}$, s.t. $\mathcal{S} = \underbrace{\mathcal{S}^0}_{\mathcal{S}^+} \oplus \underbrace{\mathcal{S}^1}_{\mathcal{S}^-} \oplus \underbrace{\mathcal{S}^2}_{\mathcal{S}^+}$

General $m=2k$:
- $\mathcal{S} = \mathbb{R}_{2k} \Box = \mathbb{R}_{k+1} \Box \cong (\mathbb{R}_{k} \otimes \mathbb{C}) \Box$ .
- $\mathcal{S}^j \coloneqq \{ X \Box + Y \Box B_1 : X, Y \in \mathbb{R}_k^j \}$
- $\mathcal{S} = \mathcal{S}^+ \oplus \mathcal{S}^-$ decomposes into two $\text{Spin}(2k)$-irreps (Weyl spinors).
- The $\mathcal{S}^j$ are $\text{SU}(k)$-irreps
  $$\begin{aligned} 
  \text{SU}(k) &= \{ s \in \text{Spin}(2k) : s \Box = \Box \} \\
  \mathfrak{su}(k) &= \{ B \in \mathbb{R}_{2k}^2 : B \Box = 0 \} 
  \end{aligned}$$

## Charge Conjugation & Majorana Spinors

Charge conjugation transforms particles into anti-particels and vice versa.

The charge conjugation operator $\mathcal{C}$ satisfies
- anti-linearity: $ \mathcal{C}(\lambda \Psi) = \bar{\lambda} \mathcal{C}(\Psi)$ where $\lambda \in \mathbb{C}$, $\forall \Psi \in \mathcal{S}$
- (anti-)idempotent: $\mathcal{C}^2 = \pm 1$
- transforms covariantly: $\mathcal{C}(\Lambda \Psi) = \Lambda\mathcal{C}( \Psi) \quad \forall \Lambda \in \text{Spin}(1,3)$

$$\Box = \tfrac{1}{2^k} \prod_{i < j} (1 - b_i b_j), \qquad b_1 \Box = b_2 \Box = \ldots = b_k \Box \ . $$
